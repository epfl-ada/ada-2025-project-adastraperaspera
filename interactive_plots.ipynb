# Interactive html plots generators

```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd
from src.scripts.visualization.interactive_plots import make_wt_tg_age_grid_scatter_from_csv,make_alignment_overlay_plot
```

```python
files = {
    "tg17": "src/data/Xenium_V1_FFPE_TgCRND8_17_9_months/normalized_counts_with_global_clusters_tg-17_with_plaque_data.csv",
    "tg5":  "src/data/Xenium_V1_FFPE_TgCRND8_5_7_months/normalized_counts_with_global_clusters_tg-5.csv",
    "tg2":  "src/data/Xenium_V1_FFPE_TgCRND8_2_5_months/normalized_counts_with_global_clusters_tg-2.csv",
    "wt13": "src/data/Xenium_V1_FFPE_wildtype_13_4_months/normalized_counts_with_global_clusters_wt-13.csv",
    "wt5":  "src/data/Xenium_V1_FFPE_wildtype_5_7_months/normalized_counts_with_global_clusters_wt-5.csv",
    "wt2":  "src/data/Xenium_V1_FFPE_wildtype_2_5_months/normalized_counts_with_global_clusters_wt-2.csv",
}
```

```python
x_col, y_col = "x_centroid", "y_centroid"
df_tg17 = pd.read_csv(files["tg17"])

df_tg5 = pd.read_csv(files["tg5"])
df_tg5_flip = df_tg5.copy()
df_tg5_flip[x_col] = -df_tg5_flip[x_col]

dx = df_tg17[x_col].mean() - df_tg5_flip[x_col].mean()
dy = df_tg17[y_col].mean() - df_tg5_flip[y_col].mean()

df_tg5_aligned = df_tg5_flip.copy()
df_tg5_aligned[x_col] += dx
df_tg5_aligned[y_col] += dy

fig = make_alignment_overlay_plot(
    df_ref=df_tg17,
    df_aligned=df_tg5_aligned,
    ref_label="TG17 (reference)",
    aligned_label="TG5 aligned → TG17",
    swap_xy=False,      # revert x/y as requested
    reverse_y=True,
    title="Alignment of Tg 5.7 months old mouse onto Tg 17.9 months old mouse",
    filename="tg5_to_tg17_alignment.html",
)
fig.show()

```

```python

age_map = {
    "2.5": ("wt2", "tg2"),
    "5.7": ("wt5", "tg5"),
    "13+": ("wt13", "tg17"),
}

fig = make_wt_tg_age_grid_scatter_from_csv(
    csv_paths=files,
    age_map=age_map,
    title="Spatial cell maps — WT vs TG",
    filename="wt_tg_age_grid.html",
    max_points=200_000,
    reverse_y=True,     # likely needed for tissue-like coords
    swap_xy=False,      # set True if it’s truly swapped
)
fig.show()
```
